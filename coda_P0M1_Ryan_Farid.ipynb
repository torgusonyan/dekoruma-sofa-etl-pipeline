## A.2 Extract

```python
import time
import re
import pandas as pd

from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.common.by import By

# Daftar list untuk menyimpan data
nama_produk = []
harga_produk = []
diskon_produk = []
harga_diskon = []
jumlah_ukuran = []
jumlah_terjual = []
area_pengiriman = []

# Browser Driver
driver = webdriver.Chrome()

base_url = "https://www.dekoruma.com/Sofa"

# Looping page
for page in range(1, 34):
    # Page 1
    if page == 1:
        url = base_url

    # Page 2, 3, dst.
    else:
        url = f"{base_url}?page={page}"
    print(f"\nScraping page {page}...")

    driver.get(url)
    time.sleep(5)

    # Ambil source code halaman setelah memuat semua produk
    html = driver.page_source
    soup = BeautifulSoup(html,"html.parser")

    # Scrape data dari setiap produk
    boxes = soup.find_all("a", class_="link-reset") # boxes dan try digunakan untuk menyamakan jumlah baris data
    for box in boxes:
    
        # Nama Produk
        try:
            nama = box.find("div", class_="css-901oao css-cens5h r-1umadx r-1enofrn r-8ay8q1 r-1cwl3u0 r-15zivkp")
            nama_produk.append(nama.get_text())
        except AttributeError:
            nama_produk.append(None)

        # Harga Produk
        try:
            harga = box.find("div", class_="css-901oao r-1umadx r-1enofrn r-8ay8q1 r-1cwl3u0 r-142tt33")
            harga_produk.append(harga.get_text())
        except AttributeError:
            harga_produk.append(None)

        # Diskon Produk
        try:
            diskon = box.find("div", class_="css-901oao r-1umadx r-10x49cs r-b88u0q r-3ucsff r-56xrmm")
            diskon_produk.append(diskon.get_text())
        except AttributeError:
            diskon_produk.append(None)

        # Harga Diskon
        try:
            diskonto = box.find("div", class_="css-901oao r-1umadx r-a023e6 r-b88u0q r-rjixqe")
            harga_diskon.append(diskonto.get_text())
        except AttributeError:
            harga_diskon.append(None)

        # Jumlah Ukuran
        try:
            size = box.find("div", class_="css-901oao r-14lw9ot r-z2wwpe r-1p0dtai r-q93js7 r-izkp9k r-1umadx r-1enofrn r-b88u0q r-1d2f490 r-8ay8q1 r-1cwl3u0 r-5oul0u r-1jkjb r-iphfwy r-uekwnc r-1h8ys4a r-u8s1d")
            jumlah_ukuran.append(size.get_text())
        except AttributeError:
            jumlah_ukuran.append(None)

        # Jumlah Terjual
        try:
            sold = box.find("div", class_="css-901oao r-1umadx r-1gkfh8e r-8ay8q1 r-1f529hi")
            jumlah_terjual.append(sold.get_text())
        except AttributeError:
            jumlah_terjual.append(None)

        # Area Pengiriman
        try:
            location = box.find("div", class_="css-901oao r-1umadx r-1enofrn r-b88u0q r-8ay8q1 r-1cwl3u0")
            area_pengiriman.append(location.get_text())
        except AttributeError:
            area_pengiriman.append(None)

# Menutup browser setelah selesai scraping
driver.quit()

# Cek Hasil
print("Nama:", len(nama_produk))
print("Harga:", len(harga_produk))
print("Diskon:", len(diskon_produk))
print("Harga Diskon:", len(harga_diskon))
print("Jumlah Ukuran:", len(jumlah_ukuran))
print("Jumlah Terjual:", len(jumlah_terjual))
print("Area Pengiriman:", len(area_pengiriman))

# DataFrame
df = pd.DataFrame({
    "nama produk": nama_produk,
    "harga produk": harga_produk,
    "diskon produk": diskon_produk,
    "harga diskon": harga_diskon,
    "jumlah ukuran": jumlah_ukuran,
    "jumlah terjual": jumlah_terjual,
    "area pengiriman": area_pengiriman
    })

df
```

## A.3 Export

```python
df.to_csv("dekoruma_raw.csv")
```

### B.1.3 Manipulasi Data

```python
# Membuang baris kalau kolom "nama produk" memiliki nilai NaN
df_bersih = df.dropna(subset=["nama produk"])
print(len(df), "->", len(df_bersih))
```

```python
# Membuang karakter non-angka untuk "harga produk"
df_bersih["harga produk"] = (df_bersih["harga produk"].str.replace(r"\D", "", regex=True))

# Membuang karakter non-angka untuk "diskon produk"
df_bersih["diskon produk"] = (df_bersih["diskon produk"].str.replace(r"\D", "", regex=True))

# Membuang karakter non-angka untuk "harga produk"
df_bersih["harga diskon"] = (df_bersih["harga diskon"].str.replace(r"\D", "", regex=True))

# Membuang karakter non-angka untuk "jumlah ukuran"
df_bersih["jumlah ukuran"] = (df_bersih["jumlah ukuran"].str.replace(r"\D", "", regex=True))

# Membuang karakter non-angka untuk "jumlah terjual"
df_bersih["jumlah terjual"] = (df_bersih["jumlah terjual"].str.replace(r"\D", "", regex=True))

print(df_bersih)
```

```python
# Merubah tipe data "harga produk" dari string ke integer
df_bersih["harga produk"] = df_bersih["harga produk"].astype("Int64")

# Merubah tipe data "harga produk" dari string ke integer
df_bersih["diskon produk"] = df_bersih["diskon produk"].astype("Int64")

# Merubah tipe data "harga produk" dari string ke integer
df_bersih["harga diskon"] = df_bersih["harga diskon"].astype("Int64")

# Merubah tipe data "harga produk" dari string ke integer
df_bersih["jumlah ukuran"] = df_bersih["jumlah ukuran"].astype("Int64")

# Merubah tipe data "harga produk" dari string ke integer
df_bersih["jumlah terjual"] = df_bersih["jumlah terjual"].astype("Int64")

df_bersih.info()
```

```python
# Menambahkan kolom "id"
df_bersih.insert(0, "id", range(1, 1 + len(df_bersih)))
```
